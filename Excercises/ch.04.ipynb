* **4.1 Modify the second code cell in Section 4.1.3 so that the function tells time for
five consecutive seconds.**

```python
import time

start=time.time()
def clock():
    time_now=time.strftime("%H:%M:%S")
    print(f"The current time is {time_now}")
    time.sleep(1)
    if time.time()-start<5:
        clock()
clock()
```
